```python

# Python Data Processing
# Veda Technology Internship - Day 9

import pandas as pd

# Step 1: Create student records
data = [
    {"Name": "Arun", "Department": "MCA", "Marks": 85},
    {"Name": "Priya", "Department": "MCA", "Marks": 92},
    {"Name": "Kavin", "Department": "BCA", "Marks": 68},
    {"Name": "Divya", "Department": "BCA", "Marks": 45},
    {"Name": "Ravi", "Department": "MCA", "Marks": 32},
    {"Name": "Meena", "Department": "BSc", "Marks": 76},
]

df = pd.DataFrame(data)

# Step 2: Create a function to process records
def process_records(records):
    records = records.copy()
    records["Result"] = records["Marks"].apply(
        lambda mark: "Pass" if mark >= 40 else "Fail"
    )
    records["Grade"] = records["Marks"].apply(
        lambda mark: (
            "A" if mark >= 80 else
            "B" if mark >= 60 else
            "C" if mark >= 40 else
            "F"
        )
    )
    return records

# Step 3: Process student records
processed_df = process_records(df)

print("PROCESSED STUDENT RECORDS")
print(processed_df.to_string(index=False))

# Step 4: Calculate summary statistics
print("\nSUMMARY STATISTICS")
print("Total Students:", len(processed_df))
print("Average Marks:", round(processed_df["Marks"].mean(), 2))
print("Highest Marks:", processed_df["Marks"].max())
print("Lowest Marks:", processed_df["Marks"].min())
print("Pass Count:", (processed_df["Result"] == "Pass").sum())
print("Fail Count:", (processed_df["Result"] == "Fail").sum())

# Step 5: Department-wise statistics
print("\nDEPARTMENT-WISE AVERAGE MARKS")
print(processed_df.groupby("Department")["Marks"].mean().round(2))

# Step 6: Display processed record examples
print("\nEXAMPLE OF PROCESSED RECORDS")
print(processed_df.head(3).to_string(index=False))
```
